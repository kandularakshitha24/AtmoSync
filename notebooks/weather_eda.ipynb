```python
import pandas as pd

df = pd.read_csv("../data/weather_data.csv")

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
import matplotlib.pyplot as plt

print("Matplotlib is ready!")
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["time"], df["temperature_2m"])

plt.title("Temperature Over Time")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["time"], df["relative_humidity_2m"])

plt.title("Humidity Over Time")
plt.xlabel("Time")
plt.ylabel("Relative Humidity (%)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["time"], df["precipitation"])

plt.title("Precipitation Over Time")
plt.xlabel("Time")
plt.ylabel("Precipitation")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["time"], df["wind_speed_10m"])

plt.title("Wind Speed Over Time")
plt.xlabel("Time")
plt.ylabel("Wind Speed")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

correlation = df[
    ["temperature_2m",
     "relative_humidity_2m",
     "precipitation",
     "wind_speed_10m"]
].corr()

print("Correlation Matrix:")
print(correlation)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Weather Variables Correlation Heatmap")
plt.tight_layout()

plt.show()
```
